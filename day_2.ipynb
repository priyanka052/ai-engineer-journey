Matplotlib

```python
#Line plot

import matplotlib.pyplot as plt

subjects = ["Python", "ML", "DL"]
marks = [85, 78, 92]

plt.plot(subjects, marks)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "subjects": ["Python", "ML", "DL"],
    "marks": [85, 78, 92]
}

df = pd.DataFrame(data)
print(df['marks'].mean())
top_marks = df['marks'].max()
print(top_marks)

print(df[df['marks'] == top_marks])

import matplotlib.pyplot as plt

subjects = ["Python", "ML", "DL"]
marks = [85, 78, 92]

plt.bar(subjects, marks)
plt.show()
```
